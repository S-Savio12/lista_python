Exercicio 03: Verificação de estoque crítico

```python
estoque_produtos = [
    "monitor", "teclado", "mouse", "headset", "gabinete"
]
estoque_quantidades = [5, 12, 2, 8, 15]

for indice in range(len(estoque_produtos)):
    produto = estoque_produtos[indice]
    quantidade = estoque_quantidades[indice]

    if quantidade < 8:
        print(
            f"ALERTA: O produto {produto} está com apenas "
            f"{quantidade} unidades no estoque!"
        )
```

Exercicio 04: Análise de custos mensais

```python
metas = {"jan": 1000, "fev": 1200, "mar": 1100}
gastos = {"jan": 900, "fev": 1350, "mar": 1100}

for mes in metas:
    if gastos[mes] <= metas[mes]:
        print(f"Mês {mes}: Dentro do orçamento.")
    else:
        diferenca = gastos[mes] - metas[mes]
        print(
            f"Mês {mes}: Orçamento estourado em "
            f"R$ {diferenca:.2f}."
        )
```

Exercicio 05: Reajuste geral de preços

```python
precos = {"celular": 1500, "tablet": 2500, "notebook": 5000}

percentual = float(
    input("Digite o percentual de aumento: ").replace(",", ".")
)

for produto in precos:
    precos[produto] = round(
        precos[produto] * (1 + percentual / 100), 2
    )

print("Catálogo atualizado:")

for produto, preco in precos.items():
    print(f"{produto}: R$ {preco:.2f}")
```

Exercicio 02: Calculadora de ISS

```python
def calcular_iss(valor_servico):
    if valor_servico > 5000:
        return valor_servico * 0.05
    return valor_servico * 0.03


print(f"ISS da nota de R$ 8.000: R$ {calcular_iss(8000):.2f}")
print(f"ISS da nota de R$ 2.000: R$ {calcular_iss(2000):.2f}")
```

Exercicio 03: Analisador de margem de lucro

```python
def analisar_margem(faturamento, custo):
    if faturamento <= 0:
        return "O faturamento deve ser maior que zero."

    lucro = faturamento - custo
    margem = lucro / faturamento

    if margem >= 0.30:
        return "Margem Saudável"
    return "Margem Baixa"


print(analisar_margem(10000, 6000))
```

Exercicio 04: Verificador de meta de vendedores

```python
def quem_bateu_meta(vendedores, meta):
    for nome, faturamento in vendedores.items():
        if faturamento >= meta:
            print(f"Vendedor {nome} bateu a meta!")


equipe_vendas = {
    "João": 12000,
    "Maria": 9500,
    "Ricardo": 10000,
    "Fernanda": 15200,
    "Paulo": 5000
}

meta_objetivo = 10000

quem_bateu_meta(equipe_vendas, meta_objetivo)
```

Exercicio 05: Conversor de moeda

```python
def converter_para_real(preco_dolar, cotacao):
    return preco_dolar * cotacao


def processar_lista_precos(precos_dolar, cotacao):
    for preco in precos_dolar:
        valor_real = converter_para_real(preco, cotacao)
        print(f"O item custa R$ {valor_real:.2f}")


precos_usd = [100, 50, 250]

processar_lista_precos(precos_usd, 5.20)
```

Exercicio 02: Resumo de folha de pagamento

```python
def calcular_folha(salario_bruto):
    desconto = salario_bruto * 0.10
    salario_liquido = salario_bruto - desconto
    return desconto, salario_liquido


desconto, salario_liquido = calcular_folha(5000)

print(
    f"Desconto: R$ {desconto:.2f} | "
    f"Salário Líquido: R$ {salario_liquido:.2f}"
)
```

Exercicio 04: Performance de vendas regionais

```python
def analisar_vendas(vendas):
    total = sum(vendas)
    media = total / len(vendas) if vendas else 0
    return total, media


dados_filiais = {
    "Matriz": [10000, 15000, 20000],
    "Filial Sul": [5000, 7000]
}

for nome, vendas in dados_filiais.items():
    total, media = analisar_vendas(vendas)

    print(
        f"Filial {nome} -> Total: R$ {total:.2f}, "
        f"Média: R$ {media:.2f}"
    )
```

Exercicio 05: Gestão de chamados de suporte

```python
def resumo_chamados(tempos):
    quantidade = len(tempos)
    tempo_maximo = max(tempos) if tempos else 0
    return quantidade, tempo_maximo


tempos = [15, 45, 10, 120, 30]

quantidade, tempo_maximo = resumo_chamados(tempos)

print(f"Quantidade de chamados: {quantidade}")
print(f"ALERTA: O tempo máximo de espera foi de {tempo_maximo} minutos!")
```
