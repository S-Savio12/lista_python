Exercício 03: "Divisão de Cargas" 

```python
total_caixas  = 1250
capacidade_caminhao  = 12

caminhao_cheio = total_caixas // capacidade_caminhao
caixas_restantes = total_caixas % capacidade_caminhao

print(f"\nCaminhoes totalmente cheios: {caminhao_cheio}")
print(f"Caixas restantes: {caixas_restantes}")


```

Exercício 04: "Analise de Margem de Lucro"  

```python
faturamento = 15000
custo_fixos = 5000
percentual_impostos = 0.15

imposto = faturamento * percentual_impostos 
lucro_liquido = faturamento - custo_fixos - imposto
margem_lucro = lucro_liquido / faturamento
meta_atingida = margem_lucro >= 0.30

print(f"\nImposto: R$ {imposto:.2f}")
print(f"Lucro Líquido: R$ {lucro_liquido:.2f}")
print(f"Margem de Lucro: {margem_lucro:.2%}")
print(f"Meta Atingida: {meta_atingida}")
```

Exercício 05: "Conversao de Tempo de Contrato" 

```python
total_meses = 40
anos = total_meses // 12
meses = total_meses % 12

print(f"\n0 contrato tem duracao de {anos} anos e {meses} meses.")
```

LISTA 03: EXERCICIO DE INPUTS 

Exercício 01: "Calculadora de Imposto Sobre Vendas" 

```python
valor_texto = input("\nDigite um valor bruto da nota: ")

valor_texto = valor_texto.replace("R$", "")
valor_texto = valor_texto.replace(".", "")
valor_texto = valor_texto.replace(",", ".")
valor_nota = float(valor_texto.strip())

imposto = valor_nota * 0.15

print(f"Valor do imposto: R$ {imposto:.2f}")
```

Exercício 02: Cadastro de Colaborador

Exercício 03: Análise de Metas de Vendas 

```python
faturamento_a = float(input("faturamento da loja A: "))
faturamento_b = float(input("faturamento da loja B: "))

faturamento_total = faturamento_a + faturamento_b
media_faturamento = faturamento_total / 2

print(f"Faturamento total das lojas: R$ {faturamento_total:.2f}")
print(f"Média de faturamento: R$ {media_faturamento:.2f}")
```
